```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python

data = {
    'Meio de Transporte': [
        'A pé',
        'Bicicleta',
        'Motocicleta ou Mototaxi',
        'Automóvel, taxi ou assemelhados',
        'Transporte coletivo',
        'Outros',
    ],
    'Branca': [19.4, 3.2, 9.9, 41.8, 25.2, 0.5],
    'Preta ou Parda': [24.8, 5.2, 14.2, 20.6, 34.6, 0.6],
    'Indígena': [37.5, 5.9, 15.4, 15.2, 22.4, 3.6],
}
df = pd.DataFrame(data)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
cores = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0', '#ffb3e6']
grupos = ['Branca', 'Preta ou Parda', 'Indígena']

for i, col in enumerate(grupos):
    axes[i].pie(
        df[col],
        labels=df['Meio de Transporte'],
        autopct='%1.3f%%',  # Colocado em 3 casas decimais
        startangle=180,  # Ângulo inicial de rotação
        colors=cores,
        wedgeprops={'edgecolor': 'white', 'linewidth': 1.2},
    )
    axes[i].set_title(f'Cor/Raça: {col}', fontsize=12, fontweight='bold')

plt.suptitle(
    'Distribuição de Mulheres por Meio de Transporte e Cor/Raça no Brasil (2022)',
    fontsize=14,
    fontweight='bold',
)
plt.tight_layout()
plt.savefig('grafico_transporte.png', dpi=300)
plt.show()
```
