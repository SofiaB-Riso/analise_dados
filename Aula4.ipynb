```python
import pandas as pd

# 4 -
df = pd.read_csv('nome_do_seu_arquivo.csv', skiprows=4)

df = df.dropna(how='all', axis=1)
df = df.dropna(subset=['UF', 'NOME DO MUNICÍPIO'])

col_uf = 'UF'
col_municipio = 'NOME DO MUNICÍPIO'
col_2010 = 'População 2010 (Alterações de Limites até 2022)1' 
col_2022 = 'População Censo 2022'

df[col_2010] = pd.to_numeric(df[col_2010], errors='coerce').fillna(0).astype(int)
df[col_2022] = pd.to_numeric(df[col_2022], errors='coerce').fillna(0).astype(int)

# 5 - 
df_estado = df.groupby(col_uf)[[col_2010, col_2022]].sum().reset_index()

# 6 - 
df_estado['Diferenca_Crescimento'] = df_estado[col_2022] - df_estado[col_2010]
df_estado_ordenado = df_estado.sort_values(by='Diferenca_Crescimento', ascending=False)

# 7 - 
df_estado_ordenado.to_csv('populacao_agregada_estado.csv', index=False, encoding='utf-8-sig')
print("Tabela por estado (ordenada por crescimento) salva com sucesso!")

# 8 & 9 - 
df_municipio = df[[col_uf, col_municipio, col_2010, col_2022]].copy()
df_municipio['Diferenca_Crescimento'] = df_municipio[col_2022] - df_municipio[col_2010]
df_municipio_ordenado = df_municipio.sort_values(by='Diferenca_Crescimento', ascending=False)


# 10 - 
df_municipio_ordenado.to_csv('populacao_por_municipio_ordenada.csv', index=False, encoding='utf-8-sig')
print("Tabela por município (ordenada por crescimento) salva com sucesso!")
```
