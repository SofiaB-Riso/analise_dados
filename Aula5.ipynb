```python
import io
import pandas as pd
```

```python
df = pd.read_csv('nome_do_seu_arquivo.csv', skiprows=4, decimal=",")
```

```python
df = df.dropna(how="all", axis=1)

print("--- 3. Estados ordenados por maior IDH em 2024 ---")
df_2024 = df[["Estado", "2024"]].sort_values(by="2024", ascending=False).reset_index(drop=True)
df_2024.index += 1  # Ajusta o índice para começar em 1º
print(df_2024)
print("\n" + "="*50 + "\n")
```

```python
df["Melhora"] = df["2024"] - df["1991"]
estado_maior_melhora = df.sort_values(by="Melhora", ascending=False).iloc[0]
print(f"Estado: {estado_maior_melhora['Estado']}")
print(f"IDH 1991: {estado_maior_melhora['1991']:.3f} -> IDH 2024: {estado_maior_melhora['2024']:.3f}")
print(f"Evolução absoluta: +{estado_maior_melhora['Melhora']:.3f}")
print("\n" + "="*50 + "\n")
```

```python
pioraram = df[df["Melhora"] < 0]
if pioraram.empty:
    print("Não! No acumulado histórico de 1991 a 2024, todos os estados melhoraram seu IDH.")
else:
    print("Estados que pioraram no período:")
    print(pioraram[["Estado", "1991", "2024"]])
print("\n" + "="*50 + "\n")
```

```python
df = df.drop(columns=["Melhora"])

print("--- 5 (Parte 2). Passando a tabela para o formato LONGO (melt) ---")
anos = [c for c in df.columns if str(c).isdigit()]
id_vars = [c for c in df.columns if c not in anos]

df_longo = df.melt(
    id_vars=id_vars,
    value_vars=anos,
    var_name="Ano",
    value_name="IDH"
)
```

```python
df_longo["Ano"] = df_longo["Ano"].astype(int)
df_longo["IDH"] = pd.to_numeric(df_longo["IDH"], errors="coerce")

print("Visualização das primeiras linhas do df_longo:")
print(df_longo.head())
```
